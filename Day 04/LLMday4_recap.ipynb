## Recap
#### 01. What is an Autoregressive model?
   - An autoregressive (AR) model is a statistical model that predicts a future value using its own previous values.

#### 02. In transformer block what is the role of LM Head?
   - Transformer block/model, the LM Head (Language Model Head) converts the model's final output into scores for possible next tokens (words or parts of words).

#### 03. Differentiate Pre-training and fine tuning
   - Pre-training: The model learns general language patterns from a large amount of data.
   - Fine-tuning: The pre-trained model is trained further on specific data to improve its performance for a particular task or purpose.
  (Pre-training teaches a model general language representations using large-scale datasets, while fine-tuning adapts an already pre-trained model to a specific task using task-specific data.)

#### 04. Describe 3 objectives in Pre-training
   
   1) Causal Language Modeling (CLM)
- Predicts the next token based on the previous tokens.
- Used in autoregressive models such as GPT.
- Example: “I love eating ___” → the model predicts “pizza”.
   2) Masked Language Modeling (MLM)
- Randomly masks some tokens in a sentence, and the model predicts the missing tokens using the surrounding context.
- Used in models such as BERT.
- Example: “I love [MASK]” → the model predicts “pizza”.
   3) Replaced Token Detection (RTD)
- The model learns to identify whether each token in a sentence is an original token or has been replaced by another token.
- Used in models such as ELECTRA.
- Example: “I love burgers” → the model identifies whether “burgers” is an original or replaced token.

#### 05. Differentiate BERT and GPT
   - BERT = Understands text using context from both directions.
         (Uses an encoder-only Transformer architecture.)
         
   - GPT = Generates text by predicting the next token.(Use decorders)
