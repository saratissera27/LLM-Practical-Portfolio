## LLM - DAY 02

#### 01). What is the difference between Natural Language Processing and Natural Language Understanding(NLU)?

   Natural Language Processing (NLP) is the broader field of AI that deals with enabling computers to work with human language.

   Natural Language Understanding (NLU) is a part of NLP that focuses specifically on helping computers understand the meaning and intention behind human language.

#### 02). What is tokenization? Give example.

  - A clear example of tokenization is replacing a real credit card number (e.g., 4111 1111 1111 1111) with a non‑sensitive token such as tok_c4a9f8b3-2104-439b-926f-f94b1fe55c2f, which can be used in systems without exposing the actual card data.
  - Data tokenization: Replaces sensitive information with meaningless tokens stored separately in a secure vault.

#### 03). What is the difference between a word, sentence and a subword token?
          - Word Tokenisation: It splits the text into individual words.
                   > Separates text based on spaces
                   > Each word is treated as one unit
                   > Does not break words further
          - Sentence Tokenization: This splits the text into individual sentences.
                   > Breaks text using punctuation (like . ? !)
                   > Helps in dividing large text into smaller parts
                   > Useful for summarization and analysis
          - Subword Tokenization: It works by splitting words into smaller meaningful parts.
                   > Breaks long or complex words into smaller pieces
                   > Helps handle unknown words
                   > Balances word and character levels

#### 04). How are Word2Vec and traditional Bag-of-words different?
          - Both Word2Vec and Bag-of-Words (BoW) convert text into numerical representations, but they do it differently.
          - Bag‑of‑Words creates sparse count‑based vectors with no understanding of meaning, while Word2Vec learns dense vectors (typically 100–300 dimensions) that capture semantic similarity by predicting context.
          example:
          - sentence 1:  "The movie was good.”
          - sentence 2:  “The movie was excellent.”

- With BoW, count the words in each sentence and get good and excellent are simply two different columns
- With Word2Vec,  learns vector representations from surrounding words,each word gets a vector. so the good and excellent get similar vector representations.


#### 05). What is NeuralNLP?
          - Neural NLP (Neural Natural Language Processing) is an approach to Natural Language Processing that uses neural networks and deep learning to understand, process, and generate human language.
- Simple real-world examples
  - Chatbots → understand and respond to user messages.
  - Sentiment analysis → determine whether a review is positive or negative.
  - Machine translation → translate English into Sinhala, French, etc.
  - Text classification → classify emails as spam or not spam.
  - Text generation → generate human-like text.
